### **Создание графика plot:**
```python
plt.plot(x, y, color='red', linewidth=1, markersize=1, marker='p')
```
* color - цвет графика
* linewidth - толщина линии
* markersize - размер маркера
* marker - тип маркера https://matplotlib.org/stable/api/markers_api.html

https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html

````python
x = np.linspace(-1, 1, 100)
y = np.exp(x)
plt.plot(x, y, color='red', linewidth=1, markersize=1, marker='p');
````

### **Несколько графиков на одном холсте**

##### **subplot()**

Функция subplot() позволяет вам размещать несколько графиков в одной фигуре, задавая количество строк и столбцов в сетке.

```
plt.subplot(nrows, ncols, index)
```
* nrows: количество строк в сетке.
* ncols: количество столбцов в сетке.
* index: индекс текущего графика (начиная с 1).

````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [2, 3, 5, 7, 11]
y2 = [1, 4, 6, 8, 10]

plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)  # 2 строки, 1 столбец, 1-й график
plt.plot(x, y1, color='blue', marker='o')
plt.title("Линейный график")
plt.xlabel("X")
plt.ylabel("Y1")

# Второй график
plt.subplot(2, 1, 2)  # 2 строки, 1 столбец, 2-й график
plt.bar(x, y2, color='orange')
plt.title("Столбчатая диаграмма")
plt.xlabel("X")
plt.ylabel("Y2")

# Показать графики
plt.tight_layout()  # Автоматически подгоняет графики для лучшего отображения
plt.show()
````

##### **subplots()**

Функция subplots() предоставляет более удобный способ создания нескольких графиков, возвращая как объект фигуры, так и массив осей.
```
fig, axs = plt.subplots(nrows, ncols)
```
* fig: объект фигуры.
* axs: массив осей.

````python
import matplotlib.pyplot as plt

# Данные для графиков
x = [1, 2, 3, 4, 5]
y1 = [2, 3, 5, 7, 11]
y2 = [1, 4, 6, 8, 10]

# Создание фигуры и осей
fig, axs = plt.subplots(2, 1, figsize=(10, 6))

# Первый график
axs[0].plot(x, y1, color='blue', marker='o')
axs[0].set_title("Линейный график")
axs[0].set_xlabel("X")
axs[0].set_ylabel("Y1")

# Второй график
axs[1].bar(x, y2, color='orange')
axs[1].set_title("Столбчатая диаграмма")
axs[1].set_xlabel("X")
axs[1].set_ylabel("Y2")

# Показать графики
plt.tight_layout()
plt.show()
````

#### Настройка расположения графиков

Вы можете настроить расположение графиков, изменяя количество строк и столбцов, а также размеры отдельных графиков. Например, вы можете создать сетку 2x2 и оставить один график пустым:

````python
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Первый график
axs[0, 0].plot(x, y1, color='blue', marker='o')
axs[0, 0].set_title("Линейный график")

# Второй график
axs[0, 1].bar(x, y2, color='orange')
axs[0, 1].set_title("Столбчатая диаграмма")

# Третий график - пустой
axs[1, 0].axis('off')  # Отключаем оси

# Четвертый график
axs[1, 1].scatter(x, y1, color='green')
axs[1, 1].set_title("Скаттер-график")

plt.tight_layout()
plt.show()
````
