## Условные операторы

image.png

Оператор if является началом условной конструкции.

Тело условной конструкции может содержать одно или несколько выражений (строк). По завершении тела может идти следующее условие, которое начинается с оператора elif (сокращение от else if — «иначе если»). Оно проверяется только в случае, если предыдущее условие не было истинным.

Синтаксис в elif аналогичен if. Операторов elif для одного блока условного оператора может быть несколько, а может не быть совсем. Последним идёт оператор else, который не содержит условия, а выполняется, только если ни одно из предыдущих условий в if и elif не выполнилось. Оператор else не является обязательным.
```python
a = 5
if a >= 0:
   print('a >= 0')
elif 4 <= a <= 5:
   print('4 <= a <= 5')
else:
   print('a < 0')
```

Условия можно писать в строку:
```python
a = 5
print('a >= 0') if a >= 0 else print('a < 0')
```

## Классы

В Python, класс (class) является основным механизмом для создания пользовательских типов данных и объектов. Класс определяет атрибуты (свойства) и методы (функции), которые могут быть использованы для работы с объектами этого класса. Ниже приведена основная информация о классах в Python.

Определение класса:

Класс определяется с помощью ключевого слова **class**, за которым следует имя класса и блок кода, заключенный в скобки. Например:
```python
class Dog:
    pass
```

Атрибуты класса определяется внутри класса и могут быть использованы для хранения данных. Атрибуты класса определяются вне методов и доступны для всех объектов класса. Например:
```python
class Dog:
    species = "Canis familiaris"

    def __init__(self, name, age):
        self.name = name
        self.age = age
```
В этом примере определяется класс Dog с атрибутом класса species и двумя атрибутами объекта name и age, которые инициализируются в методе **__init__()**.

Методы класса:

Методы класса определяются внутри класса и могут быть использованы для выполнения различных действий с объектами этого класса. Методы определяются с помощью ключевого слова def и принимают первый параметр self, который ссылается на текущий объект. Например:
```python
class Dog:
    species = "Canis familiaris"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        return f"{self.name} says: Woof!"

    def birthday(self):
        self.age += 1
```
В этом примере определяется класс Dog с двумя методами bark() и birthday(). Метод bark() возвращает строку с именем собаки и звуком лая, а метод birthday() увеличивает возраст собаки на 1.

`````python
class Dog:
    species = "Canis familiaris"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self, a):
        return f"{self.name} {a} says: {'Woof'*a}!"

    def birthday(self):
        self.age += 1
`````

# Чтение файла в питоне

В Python, чтение файла можно осуществить с помощью функции **open(file, mode='r', buffering=-1, encoding=None, errors=None, newline=None, closefd=True, opener=None)** или **open()**, которая открывает файл и возвращает объект файла. В этом объекте можно использовать метод **read()** для чтения всего содержимого файла или метод **readline()** для чтения одной строки за раз.

[Документация](https://docs.python.org/3/library/functions.html#open)

Например, для чтения всего содержимого файла "example.txt" можно использовать следующий код:
```python
file = open("example.txt", "r")
content = file.read()
print(content)
file.close()
```
В этом примере "r" указывает на то, что файл открывается для чтения.
```
Character

Meaning

'r' - чтение (read) : Открывает файл для чтения. Если файл не существует, то попытка открыть его приведет к ошибке FileNotFoundError. Если файл уже открыт в режиме записи ("w" или "a"), то попытка открыть его в режиме чтения ("r") приведет к ошибке FileExistsError.

'w' - запись (write) : Открывает файл для записи. Если файл уже существует, то он будет перезаписан. Если файла не существует, то он будет создан. Если файл уже открыт в режиме чтения ("r"), то попытка открыть его в режиме записи ("w") приведет к ошибке FileExistsError.

'a' - добавление (append) : Открывает файл для добавления информации в конец файла. Если файла не существует, то он будет создан. Если файл уже открыт в режиме чтения ("r"), то попытка открыть его в режиме добавления ("a") приведет к ошибке FileExistsError.

't' - text mode (default) - Режим текстового режима означает, что файл будет открыт в режиме чтения или записи с автоматическим преобразованием символов новой строки в символы перевода строки, соответствующие операционной системе.

'+' - open for updating (reading and writing)
```


Также можно использовать конструкцию **with**, которая автоматически закрывает файл после завершения работы с ним:
```python
with open("example.txt", "r") as file:
    content = file.read()
print(content)
```

`````python
# code
# file = open("example.txt", "r")
with open("example.txt", "r") as file:
  content = file.read()
print(content)
`````

Также можно читать файл по строкам, используя метод readline():
```python
with open("example.txt", "r") as file:
    for line in file:
        print(line)
```

В этом примере каждая строка файла будет распечатываться отдельно.

Также можно использовать методы readlines() и iterlines() для чтения файла по строкам:
```python
with open("example.txt", "r") as file:
    lines = file.readlines()
for line in lines:
    print(line)

with open("example.txt", "r") as file:
    for line in file.iterlines():
        print(line)
```
